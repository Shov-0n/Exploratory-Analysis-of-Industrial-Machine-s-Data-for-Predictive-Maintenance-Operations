```python

#machine_ID_1 



machine_ID_1 = df.loc[df['machineID'] == 1]
machine_ID_1.to_csv('machine_ID_1.csv')


machine_ID_1 = pd.read_csv("machine_ID_1.csv")

del machine_ID_1["Unnamed: 0"]


machine_ID_1["time_diff"] = np.nan

machine_ID_1.loc[0, 'time_diff'] = 3.0 


for idx in machine_ID_1.index:
    
    machine_ID_1.at[(idx+1), 'time_diff'] = machine_ID_1.at[(idx), 'time_diff'] + 3.0 

machine_ID_1 = machine_ID_1.drop([2913])    

machine_ID_1.to_csv('machine_ID_1.csv') # save file


```
